# Evaluation demo

This notebook shows the saved results of all 11 simulations in the paper: Table 1 and the matching plots. It only reads files, so **Run > Run All Cells** takes seconds. To run the simulations yourself, open `run_demo.ipynb`.


```python
import sys
sys.path.insert(0, "..")  # lets the notebook find the code in this repository

import csv
from pathlib import Path

from IPython.display import Image, Markdown, display

from simulations import configs
from simulations.run import slug

RESULTS = Path("..") / "results"
```

## Table 1

Each value is the mean ± SD of 10 repeats. The last column is the value printed in the current paper, which came from the 2021 code.


```python
display(Markdown((RESULTS / "table1.md").read_text()))
```

## Matching plots of each simulation

Each simulation shows the mean metrics of its 10 repeats, then the true matching and the solver's matching from its last repeat. Each line connects a cell to its match, and colours show the true clusters.


```python
with open(RESULTS / "summary.csv") as handle:
    summary = {row["simulation"]: row for row in csv.DictReader(handle)}

for label, name in configs.PAPER_SIMULATIONS.items():
    row = summary[name]
    display(Markdown(f"### {label}: {row['description']}\n\n"
                     f"Acc {float(row['label_accuracy_mean']):.3f}, "
                     f"RMS {float(row['relaxed_match_accuracy_mean']):.3f}, "
                     f"LDE {float(row['avg_dist_solution_mean']):.3f}"))
    for kind in ("ground_truth", "solution"):
        display(Image(filename=str(RESULTS / "figures" / f"{slug(name)}_{kind}.png")))
```
